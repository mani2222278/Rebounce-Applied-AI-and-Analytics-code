### Exercise 6 — Reshape a 1D array into 2D

### Exercise 7 — Inspect an array's properties

Given the array below, print its shape, data type (`dtype`), total number of elements (`size`), and number of dimensions (`ndim`):

```python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
```

````python
# Write your code here
data = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("Shape:", data.shape)
print("Dtype:", data.dtype)
print("Size:", data.size)
print("Dimensions:", data.ndim)
````

### Exercise 8 — Add and subtract arrays element-wise

Create two arrays, `prices = [100, 250, 400]` and `discounts = [10, 25, 50]`. Print the element-wise sum of the two arrays, and the result of subtracting `discounts` from `prices`.

````python
# Write your code here
prices = np.array([100, 250, 400])
discounts = np.array([10, 25, 50])

print("Sum:", prices + discounts)
print("After discount:", prices - discounts)
````

### Exercise 9 — Multiply arrays and scale by a number

Create an array `quantities = [2, 5, 3]` and an array `unit_prices = [50, 20, 100]`. Print the element-wise product (the total cost per item). Then print the result of multiplying `quantities` by the scalar `10`.

````python
# Write your code here
quantities = np.array([2, 5, 3])
unit_prices = np.array([50, 20, 100])

print("Total cost:", quantities * unit_prices)
print("Quantities × 10:", quantities * 10)
````
