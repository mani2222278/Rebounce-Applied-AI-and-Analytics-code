### Exercise 11 — Challenge — build objects and collect their data

Define an `Employee` class with attributes `name`, `department`, and `salary`. Create four Employee objects with different values, then collect their data into a list of dictionaries (one dictionary per employee).

````python
# Write your code here
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary


emp1 = Employee("Rahul", "IT", 50000)
emp2 = Employee("Priya", "HR", 45000)
emp3 = Employee("Amit", "Finance", 55000)
emp4 = Employee("Neha", "Marketing", 48000)


employees = [
    {"name": emp1.name, "department": emp1.department, "salary": emp1.salary},
    {"name": emp2.name, "department": emp2.department, "salary": emp2.salary},
    {"name": emp3.name, "department": emp3.department, "salary": emp3.salary},
    {"name": emp4.name, "department": emp4.department, "salary": emp4.salary}
]

print(employees)
````

---

### Exercise 12 — Challenge — turn objects into a DataFrame

Convert the list of employee dictionaries from the previous exercise into a pandas DataFrame and display it. Then print the average salary across all four employees.

````python
# Write your code here
employee_df = pd.DataFrame(employees)

print(employee_df)

print("Average salary:", employee_df["salary"].mean())
````
