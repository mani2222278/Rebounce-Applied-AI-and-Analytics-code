 textTrack A - Personal Expense Tracker

This program help users record their daily expenses.
It shows total expensse and category-wise total.
Run all cells, then call main() to start.


```python
import json
import os

FILE_NAME = "expenses.json"

# Load expenses
if os.path.exists(FILE_NAME):
    with open(FILE_NAME, "r") as file:
        expenses = json.load(file)
else:
    expenses = []


def add_expense():
    category = input("Enter category: ")
    amount = float(input("Enter amount: "))
    description = input("Enter description: ")

    expense = {
        "category": category,
        "amount": amount,
        "description": description
    }

    expenses.append(expense)
    print("Expense added successfully!\n")


def view_expenses():
    if not expenses:
        print("No expenses found.\n")
        return

```

```python
 print("\n===== All Expenses =====")
for i, e in enumerate(expenses,start=1):
        print(f"{i}. Category    : {e['category']}")
        print(f"   Amount      : Rs {e['amount']:.2f}")
        print(f"   Description : {e['description']}")
        print()


def total_spent():
    total = sum(e["amount"] for e in expenses)
    print(f"\nTotal Amount Spent: Rs {total:.2f}\n")


def category_totals():
    totals = {}

    for e in expenses:
        category = e["category"]
        totals[category] = totals.get(category, 0) + e["amount"]

    print("\n===== Totals by Category =====")
    for category, amount in totals.items():
        print(f"{category}: Rs {amount:.2f}")

    total = sum(totals.values())
    print(f"Total: Rs {total:.2f}\n")


def save_data():
    with open(FILE_NAME, "w") as file:
        json.dump(expenses, file, indent=4)
    print("Data saved successfully!")
```

```python

while True:
    print("\n===== Expense Tracker =====")
    print("1. Add Expense")
    print("2. View All Expenses")
    print("3. View Total Amount")
    print("4. View Totals by Category")
    print("5. Save and Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_expense()

    elif choice == "2":
        view_expenses()

    elif choice == "3":
        total_spent()

    elif choice == "4":
        category_totals()

    elif choice == "5":
        save_data()
        print("Thank you for using Expense Tracker!")
        break

    else:
        print("Invalid choice! Please try again.")
```
