```python
scores = {
    "Anika": 35,
    "Rohan": 72,
    "Meera": 40,
    "Kabir": 28,
    "Diya":  88,
    "Ishaan": 55,
    }
passing_filter = filter(lambda item: item[1] >= 40, scores.items())
passing_students = [item[0] for item in passing_filter]
print(f"Passing students:{passing_students}")
```
