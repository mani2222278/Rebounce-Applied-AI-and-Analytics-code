## Section 2 · Conditionals — if / elif / else

Up to now your programs have run every line, top to bottom, no matter what. A **conditional** lets your program make a decision — run some code only **if** something is true.

- `if` — run a block when a condition is `True`
- `elif` — check another condition if the first was `False` (short for "else if")
- `else` — the fallback, runs when nothing above matched

Python uses **indentation** (4 spaces) to know what belongs inside the block — not curly braces.


### 2.1 · if / else

````python
age = 20

if age >= 18:
    print("Adult")
else:
    print("Minor")
# Expected: Adult
````

### 2.2 · if / elif / else — more than two outcomes

````python
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Grade:{grade}")
# Expected: Grade: C
````

````python
print("Grade:",grade)
````

````python
age    = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
else:
    print("Entry denied")
# Expected: Entry allowed
````

## Section 3 · Loops — for, while, break, continue

A **loop** repeats a block of code. Python gives you two kinds:

- `for` — repeat once **for each item** in something (a list, a string, a range of numbers)
- `while` — repeat **as long as** a condition stays `True`

Inside a loop, `break` stops it completely, and `continue` skips straight to the next round.


### 3.1 · for loop over a list

````python
fruits = ["apple", "banana", "mango"]

for x in fruits:
    print(x)
# Expected:
# apple
# banana
# mango
````

### 3.2 · range() and enumerate()

````python
# range(5) counts 0,1,2,3,4
for i in range(5):
    print(i)

print("---")

# enumerate() gives you the index AND the item together
fruits = ["apple", "banana", "mango"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

### 3.3 · while loop

````python
count = 1

while count <= 10:
    print(count)
    count += 1   # without this line, the loop never ends!
````

### 3.4 · break and continue

````python
# break — stop the loop completely
for num in range(1, 10):
    if num == 5:
        break
    print(num)

print("---")

# continue — skip this round, keep going
for num in range(1, 10):
    if num % 2 == 0:
        continue
    print(num)
````

## Section 4 · Functions

A **function** packages a piece of logic under a name, so you can reuse it instead of copy-pasting code. You've already been *using* functions all along (`print`, `len`, `sorted`) — now you'll write your own.


### 4.1 · Defining and calling a function

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Asha")
greet("Rahul")
````

### 4.2 · return values

`print()` just displays something. `return` sends a value **back**, so you can store it in a variable and use it later.


````python
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
# Expected: 8
````

### 4.5 · Variable scope — local vs global

A variable created **inside** a function only exists inside that function (its **local scope**). It doesn't affect a variable with the same name outside.


````python
x = 10   # global
def show():
    x = 5   # local, separate from the global x
    print("Inside function:", x)
show()

print(x)

````

## Section 5 · Handling errors — try / except

Real data is messy — a number field might contain text, a file might be missing. Without handling, one bad value crashes your whole program. `try` / `except` lets you catch the error and decide what to do instead of crashing.


### 5.1 · Basic try / except

````python
try:
    num = int("abc")
except ValueError:
    print("That's not a valid number")
````

### 5.2 · Catching different error types

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Can't divide by zero")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

### 5.3 · else and finally

````python
try:
    num = int("42")
except ValueError:
    print("Invalid input")
else:
    print("Conversion worked:", num)   # runs only if no error happened
finally:
    print("Done checking input")        # always runs
````

## Section 5.4 · Input validation with a loop

`try/except` handles bad data you already have. But often you're **asking the user** for input, and you want to keep asking until they give you something valid — that's what a `while True` loop is for.

- Loop forever with `while True`
- Check the input; if it's valid, `break` out of the loop
- If not, print a message and loop again automatically

````python
while True:
    age = input("Enter your age: ")

    if age.isdigit():
        age = int(age)
        break

    print("Please enter a valid number.")

print(f"Your age is {age}")
````

> **Note:** this cell uses `input()`, so it will pause and wait for you to actually type a response when you run it — it won't just print output on its own like the other cells.

Test it with `[10, 20, 30]` and `[]`.

## Section 6 · Putting it together — starting MP2

MP2 is a small program that **reads a batch of raw text data, cleans it, and reports on it** — exactly the shape of a real data-entry problem. You now have every tool you need:

- a **loop** to go through each row
- an **if/elif** to classify or validate each value
- a **function** to keep the logic reusable
- **try/except** so one bad row doesn't stop the whole program


### Worked example — parsing raw text into clean records

````python
raw_data = "Asha,88\nRahul,72\nPriya,95\nAmit,abc\nSneha,45"

def parse_line(line):
    try:
        name, marks = line.split(",")
        return {"name": name, "marks": int(marks)}
    except ValueError:
        print(f"Skipping bad row: {line}")
        return None

records = []
for line in raw_data.split("\n"):
    record = parse_line(line)
    if record:
        records.append(record)

print(records)
````

### Worked example — reporting pass/fail on the clean data

````python
def status(marks):
    if marks >= 50:
        return "Pass"
    else:
        return "Fail"

for record in records:
    print(f"{record['name']}: {record['marks']} -> {status(record['marks'])}")
````

Hint: `any(char.isdigit() for char in password)` checks if a string has a digit.

### Challenge 6 · Mini CSV parser (MP2 warm-up)

You have raw `"name,marks"` rows, one per line, some with bad data. Write a function `load_students(raw_text)` that:

1. Splits the text into lines
2. Parses each line into `{"name": ..., "marks": ...}`, skipping bad rows with try/except
3. Returns the list of clean records

Then write a second function `class_average(records)` that returns the average marks, and print each student's pass/fail status (pass = marks >= 50).

```python
raw_text = "Asha,88\nRahul,72\nPriya,xx\nAmit,45\nSneha,"
```


````python
# YOUR CODE HERE

````

**Solution:**

````python
def load_students(raw_text):
    records = []
    for line in raw_text.split("\n"):
        try:
            name, marks = line.split(",")
            records.append({"name": name, "marks": int(marks)})
        except ValueError:
            print(f"Skipping bad row: {line}")
    return records
def class_average(records):
    marks = [r["marks"] for r in records]
    return sum(marks) / len(marks)

raw_text = "Asha,88\nRahul,72\nPriya,xx\nAmit,45\nSneha,"

students = load_students(raw_text)
print(f"Class average: {round(class_average(students), 1)}")

for s in students:
    result = "Pass" if s["marks"] >= 50 else "Fail"
    print(f"{s['name']}: {s['marks']} -> {result}")
````
